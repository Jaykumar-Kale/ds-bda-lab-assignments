```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------------------------------
# Define Column Names
# ---------------------------------------------------

column_names = [
    'age',
    'sex',
    'cp',
    'trestbps',
    'chol',
    'fbs',
    'restecg',
    'thalach',
    'exang',
    'oldpeak',
    'slope',
    'ca',
    'thal',
    'target'
]

# ---------------------------------------------------
# Load Dataset
# ---------------------------------------------------

df = pd.read_csv(
    "../datasets/HeartDisease/Cleavland.csv",
    header=None,
    names=column_names
)

# ---------------------------------------------------
# Display Dataset
# ---------------------------------------------------

print(df.head())

# ---------------------------------------------------
# HISTOGRAM
# Objective:
# To show frequency distribution of age
# ---------------------------------------------------

plt.figure(figsize=(6,4))

plt.hist(df['age'], bins=10)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()

# ---------------------------------------------------
# DOT PLOT
# Objective:
# To show distribution of cholesterol values
# ---------------------------------------------------

plt.figure(figsize=(8,4))

plt.plot(df['chol'], 'o')

plt.title("Dot Plot of Cholesterol")

plt.ylabel("Cholesterol")

plt.show()

# ---------------------------------------------------
# BAR PLOT
# Objective:
# To compare average cholesterol by sex
# ---------------------------------------------------

plt.figure(figsize=(6,4))

sns.barplot(
    x='sex',
    y='chol',
    data=df
)

plt.title("Average Cholesterol by Sex")

plt.show()

# ---------------------------------------------------
# LINE CHART
# Objective:
# To show trend of maximum heart rate
# ---------------------------------------------------

plt.figure(figsize=(8,4))

plt.plot(df['thalach'])

plt.title("Maximum Heart Rate Trend")

plt.ylabel("Heart Rate")

plt.show()

# ---------------------------------------------------
# BOX PLOT
# Objective:
# To detect outliers in cholesterol
# ---------------------------------------------------

plt.figure(figsize=(6,4))

sns.boxplot(
    x=df['chol']
)

plt.title("Box Plot of Cholesterol")

plt.show()

# ---------------------------------------------------
# SCATTER PLOT
# Objective:
# To show relationship between age and cholesterol
# ---------------------------------------------------

plt.figure(figsize=(6,4))

plt.scatter(
    df['age'],
    df['chol']
)

plt.title("Age vs Cholesterol")

plt.xlabel("Age")

plt.ylabel("Cholesterol")

plt.show()

# ---------------------------------------------------
# BOXPLOT + SCATTERPLOT
# Objective:
# To visualize distribution and outliers together
# ---------------------------------------------------

plt.figure(figsize=(6,4))

sns.boxplot(
    x=df['chol']
)

plt.scatter(
    range(len(df['chol'])),
    df['chol'],
    alpha=0.5
)

plt.title("Boxplot with Scatterplot")

plt.show()
```
