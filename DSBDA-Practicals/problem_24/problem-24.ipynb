```python
import pandas as pd

# ---------------------------------------------------
# Load Dataset
# ---------------------------------------------------

df = pd.read_csv("../datasets/forestfires/forestfires.csv")

# ---------------------------------------------------
# Display Dataset
# ---------------------------------------------------

print("\nFIRST 5 RECORDS")
print(df.head())

# ---------------------------------------------------
# Create Classes Based on Area Affected
# ---------------------------------------------------

df['Region'] = pd.cut(
    df['area'],
    bins=[-1, 0, 50, df['area'].max()],
    labels=['NotAffected', 'PartiallyAffected', 'MostlyAffected']
)

print("\nREGION CLASSIFICATION")
print(df[['area', 'Region']].head())

# ---------------------------------------------------
# 1. Create Subsets
# ---------------------------------------------------

partial = df[df['Region'] == 'PartiallyAffected']
mostly = df[df['Region'] == 'MostlyAffected']

print("\nPARTIALLY AFFECTED")
print(partial.head())

print("\nMOSTLY AFFECTED")
print(mostly.head())

# ---------------------------------------------------
# 2. Merge Subsets
# ---------------------------------------------------

merged_data = pd.concat([partial, mostly])

print("\nMERGED DATA")
print(merged_data.head())

# ---------------------------------------------------
# 3. Sort Data
# ---------------------------------------------------

sorted_data = df.sort_values(
    by=['temp', 'wind', 'area'],
    ascending=False
)

print("\nSORTED DATA")
print(sorted_data[['temp', 'wind', 'area']].head())

# ---------------------------------------------------
# 4. Transpose Data
# ---------------------------------------------------

transposed_data = df.head().T

print("\nTRANSPOSED DATA")
print(transposed_data)

# ---------------------------------------------------
# 5. Melt Data
# ---------------------------------------------------

melted_data = pd.melt(
    df,
    id_vars=['Region'],
    value_vars=['temp']
)

print("\nMELTED DATA")
print(melted_data.head())

# ---------------------------------------------------
# 6. Pivot Table
# ---------------------------------------------------

pivot_data = df.pivot_table(
    values='area',
    index='Region',
    aggfunc='mean'
)

print("\nPIVOT TABLE")
print(pivot_data)
```
