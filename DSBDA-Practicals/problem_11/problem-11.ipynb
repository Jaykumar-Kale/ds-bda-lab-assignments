```python
import pandas as pd

# ---------------------------------------------------
# Load Dataset
# ---------------------------------------------------

df = pd.read_csv(
    "../datasets/Movie/movies_metadata.csv",
    low_memory=False
)

# ---------------------------------------------------
# Keep Required Columns Only
# ---------------------------------------------------

df = df[['title', 'original_language', 'vote_average']]

# ---------------------------------------------------
# Remove Missing Values
# ---------------------------------------------------

df = df.dropna()

# ---------------------------------------------------
# Convert vote_average to numeric
# ---------------------------------------------------

df['vote_average'] = pd.to_numeric(
    df['vote_average'],
    errors='coerce'
)

# Remove invalid ratings
df = df.dropna()

# ---------------------------------------------------
# Display Dataset
# ---------------------------------------------------

print("\nFIRST 5 RECORDS")
print(df.head())

# ---------------------------------------------------
# 1. Create Subsets
# ---------------------------------------------------

english_movies = df[df['original_language'] == 'en']
french_movies = df[df['original_language'] == 'fr']

print("\nENGLISH MOVIES")
print(english_movies.head())

print("\nFRENCH MOVIES")
print(french_movies.head())

# ---------------------------------------------------
# 2. Merge Subsets
# ---------------------------------------------------

merged_data = pd.concat([english_movies, french_movies])

print("\nMERGED DATA")
print(merged_data.head())

# ---------------------------------------------------
# 3. Sort Data Using Ratings
# ---------------------------------------------------

sorted_data = df.sort_values(
    by='vote_average',
    ascending=False
)

print("\nSORTED DATA")
print(sorted_data.head())

# ---------------------------------------------------
# 4. Transpose Data
# ---------------------------------------------------

transposed_data = df.head().T

print("\nTRANSPOSED DATA")
print(transposed_data)

# ---------------------------------------------------
# 5. Melt Data
# ---------------------------------------------------

melted_data = pd.melt(
    df,
    id_vars=['original_language'],
    value_vars=['vote_average']
)

print("\nMELTED DATA")
print(melted_data.head())

# ---------------------------------------------------
# 6. Pivot Table
# ---------------------------------------------------

pivot_data = df.pivot_table(
    values='vote_average',
    index='original_language',
    aggfunc='mean'
)

print("\nPIVOT TABLE")
print(pivot_data)
```
