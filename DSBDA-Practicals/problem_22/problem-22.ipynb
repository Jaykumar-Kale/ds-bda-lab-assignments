```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------------------------------
# Define Column Names
# ---------------------------------------------------

column_names = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education_num',
    'marital_status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital_gain',
    'capital_loss',
    'hours_per_week',
    'native_country',
    'income'
]

# ---------------------------------------------------
# Load Dataset
# ---------------------------------------------------

df = pd.read_csv(
    "../datasets/Adult/adult.csv",
    header=None,
    names=column_names
)

# ---------------------------------------------------
# Display Dataset
# ---------------------------------------------------

print(df.head())

# ---------------------------------------------------
# HISTOGRAM
# Objective:
# To show distribution of working hours
# ---------------------------------------------------

plt.figure(figsize=(6,4))

plt.hist(
    df['hours_per_week'],
    bins=10
)

plt.title("Distribution of Working Hours")

plt.xlabel("Hours Per Week")

plt.ylabel("Frequency")

plt.show()

# ---------------------------------------------------
# PIE CHART
# Objective:
# To show income category distribution
# ---------------------------------------------------

income_count = df['income'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    income_count,
    labels=income_count.index,
    autopct='%1.1f%%'
)

plt.title("Income Distribution")

plt.show()

# ---------------------------------------------------
# BAR GRAPH
# Objective:
# To compare average age by income category
# ---------------------------------------------------

plt.figure(figsize=(6,4))

sns.barplot(
    x='income',
    y='age',
    data=df
)

plt.title("Average Age by Income")

plt.show()

# ---------------------------------------------------
# SCATTER PLOT
# Objective:
# To show relationship between age and capital gain
# ---------------------------------------------------

plt.figure(figsize=(6,4))

plt.scatter(
    df['age'],
    df['capital_gain']
)

plt.title("Age vs Capital Gain")

plt.xlabel("Age")

plt.ylabel("Capital Gain")

plt.show()

# ---------------------------------------------------
# BOX PLOT
# Objective:
# To detect outliers in capital gain
# ---------------------------------------------------

plt.figure(figsize=(6,4))

sns.boxplot(
    x=df['capital_gain']
)

plt.title("Box Plot of Capital Gain")

plt.show()

# ---------------------------------------------------
# BOXPLOT + SCATTERPLOT
# Objective:
# To visualize outliers and distribution together
# ---------------------------------------------------

plt.figure(figsize=(6,4))

sns.boxplot(
    x=df['capital_gain']
)

plt.scatter(
    range(len(df['capital_gain'])),
    df['capital_gain'],
    alpha=0.4
)

plt.title("Boxplot with Scatterplot")

plt.show()
```
