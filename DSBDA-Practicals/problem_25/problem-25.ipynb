```python
import pandas as pd

# ---------------------------------------------------
# Define Column Names
# ---------------------------------------------------

column_names = [
    'CLASS',
    'AGE',
    'SEX',
    'STEROID',
    'ANTIVIRALS',
    'FATIGUE',
    'MALAISE',
    'ANOREXIA',
    'LIVER_BIG',
    'LIVER_FIRM',
    'SPLEEN_PALPABLE',
    'SPIDERS',
    'ASCITES',
    'VARICES',
    'BILIRUBIN',
    'ALK_PHOSPHATE',
    'SGOT',
    'ALBUMIN',
    'PROTIME',
    'HISTOLOGY'
]

# ---------------------------------------------------
# Load Dataset
# ---------------------------------------------------

df = pd.read_csv(
    "../datasets/Hepatitis/hepatitis.csv",
    header=None,
    names=column_names
)

# ---------------------------------------------------
# Display Dataset
# ---------------------------------------------------

print("\nFIRST 5 RECORDS")
print(df.head())

# ---------------------------------------------------
# Create Subsets
# ---------------------------------------------------

male_data = df[df['SEX'] == 1]
female_data = df[df['SEX'] == 2]

print("\nMALE DATA")
print(male_data.head())

print("\nFEMALE DATA")
print(female_data.head())

# ---------------------------------------------------
# Merge Subsets
# ---------------------------------------------------

merged_data = pd.concat([male_data, female_data])

print("\nMERGED DATA")
print(merged_data.head())

# ---------------------------------------------------
# Sort Data
# ---------------------------------------------------

sorted_data = df.sort_values(
    by=['AGE', 'SGOT', 'PROTIME'],
    ascending=False
)

print("\nSORTED DATA")
print(sorted_data[['AGE', 'SGOT', 'PROTIME']].head())

# ---------------------------------------------------
# Transpose Data
# ---------------------------------------------------

transposed_data = df.head().T

print("\nTRANSPOSED DATA")
print(transposed_data)

# ---------------------------------------------------
# Melt Data
# ---------------------------------------------------

melted_data = pd.melt(
    df,
    id_vars=['SEX'],
    value_vars=['AGE']
)

print("\nMELTED DATA")
print(melted_data.head())

# ---------------------------------------------------
# Pivot Table
# ---------------------------------------------------

pivot_data = df.pivot_table(
    values='AGE',
    index='SEX',
    aggfunc='mean'
)

print("\nPIVOT TABLE")
print(pivot_data)
```
