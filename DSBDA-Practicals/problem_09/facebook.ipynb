Q-9 : Perform the Following operations using Python on the Facebook Metrics Dataset:

1. Create the Subset for the Type of Post
2. Merge the 2 subset
3. sort the dataset on the Page Total likes
4. Transpose the data
5. Melting the data to long format
6. Casting data to wide format.

```python
import pandas as pd
```

Load the dataset:

```python
df = pd.read_csv("../datasets/dataset_Facebook.csv", sep=";")
df.head()
```

```python
df.info()
```

1. Create the subset for the Type of post:

```python
photos_post = df[df['Type'] =='Photo']
status_post = df[df['Type']=='Status']

print("\n first 5 Photos post: \n")
print(photos_post.head())

print("\n first 5 Status post: \n")
print(status_post.head())
```

2. Merge the data

```python
merged_data = pd.concat([photos_post, status_post])
print("\n Merged Data: \n")
print(merged_data.head())
```

3. Sort the data:

```python
sorted_data = df.sort_values(by='Page total likes',ascending=False)
print("\n Sorted Data: \n")
print(sorted_data[['Type', 'Page total likes']].head())
```

4. Transpose data:

```python
transposed_data = df.head().T
```

```python
print("transposed_data : ")
print(transposed_data)
```

5. Melt data to long format

```python
melted_data = pd.melt(df, id_vars=['Type'],value_vars = ['Page total likes'])
print("\n Melted Data: \n")
print(melted_data.head())
```

6. casting the data to wide format

```python
casted_data = df.pivot_table(values='Page total likes',index='Type',aggfunc='mean')
print("\n Casted Data: \n")
print(casted_data)
```
