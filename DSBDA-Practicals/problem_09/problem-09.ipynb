```python
import pandas as pd
```

```python
df = pd.read_csv("../datasets/dataset_Facebook.csv", sep=";")
df.head()
```

```python
photo_posts = df[df['Type'] == 'Photo']
status_posts = df[df['Type'] == 'Status']

photo_posts.head()
```

```python
merged_data = pd.concat([photo_posts, status_posts])

merged_data.head()
```

```python
sorted_data = df.sort_values(by='Page total likes', ascending=False)

sorted_data[['Type', 'Page total likes']].head()
```

```python
transposed_data = df.head().T

transposed_data
```

```python
melted_data = pd.melt(
    df,
    id_vars=['Type'],
    value_vars=['Page total likes']
)

melted_data.head()
```

```python
pivot_data = df.pivot_table(
    values='Page total likes',
    index='Type',
    aggfunc='mean'
)

pivot_data
```
