```python
import pandas as pd

# ---------------------------------------------------
# Define Column Names
# ---------------------------------------------------

column_names = [
    'age',
    'workclass',
    'fnlwgt',
    'education',
    'education_num',
    'marital_status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'capital_gain',
    'capital_loss',
    'hours_per_week',
    'native_country',
    'income'
]

# ---------------------------------------------------
# Load Dataset
# ---------------------------------------------------

df = pd.read_csv(
    "../datasets/Adult/adult.csv",
    header=None,
    names=column_names
)

# ---------------------------------------------------
# Display Dataset
# ---------------------------------------------------

print("\nFIRST 5 RECORDS")
print(df.head())

# ---------------------------------------------------
# 1. Create Subsets
# ---------------------------------------------------

male_data = df[df['sex'] == ' Male']
female_data = df[df['sex'] == ' Female']

print("\nMALE DATA")
print(male_data.head())

print("\nFEMALE DATA")
print(female_data.head())

# ---------------------------------------------------
# 2. Merge Subsets
# ---------------------------------------------------

merged_data = pd.concat([male_data, female_data])

print("\nMERGED DATA")
print(merged_data.head())

# ---------------------------------------------------
# 3. Sort Data Using Age
# ---------------------------------------------------

sorted_data = df.sort_values(
    by='age',
    ascending=False
)

print("\nSORTED DATA")
print(sorted_data.head())

# ---------------------------------------------------
# 4. Transpose Data
# ---------------------------------------------------

transposed_data = df.head().T

print("\nTRANSPOSED DATA")
print(transposed_data)

# ---------------------------------------------------
# 5. Melt Data
# ---------------------------------------------------

melted_data = pd.melt(
    df,
    id_vars=['sex'],
    value_vars=['age']
)

print("\nMELTED DATA")
print(melted_data.head())

# ---------------------------------------------------
# 6. Pivot Table
# ---------------------------------------------------

pivot_data = df.pivot_table(
    values='age',
    index='sex',
    aggfunc='mean'
)

print("\nPIVOT TABLE")
print(pivot_data)
```
