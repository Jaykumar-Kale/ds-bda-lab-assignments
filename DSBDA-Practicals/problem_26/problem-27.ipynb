```python
import requests
import pandas as pd

# ---------------------------------------------------
# URL
# ---------------------------------------------------

url = "https://store.steampowered.com/appreviews/730?json=1"

# ---------------------------------------------------
# Send GET Request
# ---------------------------------------------------

response = requests.get(url)

# ---------------------------------------------------
# Convert Response to JSON
# ---------------------------------------------------

data = response.json()

# ---------------------------------------------------
# Extract Reviews
# ---------------------------------------------------

reviews = data['reviews']

# ---------------------------------------------------
# Create Empty List
# ---------------------------------------------------

review_data = []

# ---------------------------------------------------
# Loop Through Reviews
# ---------------------------------------------------

for review in reviews:

    customer_name = review['author']['steamid']

    comment = review['review']

    rating = review['voted_up']

    votes = review['votes_up']

    # Append Data
    review_data.append([
        customer_name,
        comment,
        rating,
        votes
    ])

# ---------------------------------------------------
# Create DataFrame
# ---------------------------------------------------

df = pd.DataFrame(
    review_data,
    columns=[
        'Customer_Name',
        'Comment',
        'Recommended',
        'Votes'
    ]
)

# ---------------------------------------------------
# Display Output
# ---------------------------------------------------

print(df.head())

# ---------------------------------------------------
# Save CSV File
# ---------------------------------------------------

df.to_csv("steam_reviews.csv", index=False)

print("\nData Saved Successfully")
```
