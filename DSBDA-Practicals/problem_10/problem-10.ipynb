```python
import pandas as pd

column_names = [
    'SepalLengthCm',
    'SepalWidthCm',
    'PetalLengthCm',
    'PetalWidthCm',
    'Species'
]

df = pd.read_csv(
    "../datasets/Iris/Iris.csv",
    header=None,
    names=column_names
)

df.head()
```

```python
setosa = df[df['Species'] == 'Iris-setosa']
virginica = df[df['Species'] == 'Iris-virginica']

setosa.head()
```

```python
merged_data = pd.concat([setosa, virginica])

merged_data.head()
```

```python
sorted_data = df.sort_values(by='PetalLengthCm', ascending=False)

sorted_data.head()
```

```python
transposed_data = df.head().T

transposed_data
```

```python
melted_data = pd.melt(
    df,
    id_vars=['Species'],
    value_vars=['PetalLengthCm']
)

melted_data.head()
```

```python
pivot_data = df.pivot_table(
    values='PetalLengthCm',
    index='Species',
    aggfunc='mean'
)

pivot_data
```
