```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# ---------------------------------------------------
# URL
# ---------------------------------------------------

url = "https://webscraper.io/test-sites/e-commerce/static"

# ---------------------------------------------------
# Send GET Request
# ---------------------------------------------------

response = requests.get(url)

# ---------------------------------------------------
# Parse HTML Content
# ---------------------------------------------------

soup = BeautifulSoup(response.text, 'html.parser')

# ---------------------------------------------------
# Find All Product Cards
# ---------------------------------------------------

products = soup.find_all('div', class_='thumbnail')

# ---------------------------------------------------
# Empty List to Store Data
# ---------------------------------------------------

data = []

# ---------------------------------------------------
# Loop Through Products
# ---------------------------------------------------

for product in products:

    # ---------------------------------------------------
    # Product Name
    # ---------------------------------------------------

    name_tag = product.find('a', class_='title')

    if name_tag:
        name = name_tag.text.strip()
    else:
        name = "No Name"

    # ---------------------------------------------------
    # Price
    # ---------------------------------------------------

    price_tag = product.find('h4', class_='price')

    if price_tag:
        price = price_tag.text.strip()
    else:
        price = "No Price"

    # ---------------------------------------------------
    # Description
    # ---------------------------------------------------

    desc_tag = product.find('p', class_='description')

    if desc_tag:
        description = desc_tag.text.strip()
    else:
        description = "No Description"

    # ---------------------------------------------------
    # Reviews
    # ---------------------------------------------------

    review_tag = product.find('p', class_='pull-right')

    if review_tag:
        reviews = review_tag.text.strip()
    else:
        reviews = "No Reviews"

    # ---------------------------------------------------
    # Ratings
    # ---------------------------------------------------

    rating = len(
        product.find_all(
            'span',
            class_='glyphicon-star'
        )
    )

    # ---------------------------------------------------
    # Append Data
    # ---------------------------------------------------

    data.append([
        name,
        price,
        description,
        reviews,
        rating
    ])

# ---------------------------------------------------
# Create DataFrame
# ---------------------------------------------------

df = pd.DataFrame(
    data,
    columns=[
        'Product_Name',
        'Price',
        'Description',
        'Reviews',
        'Rating'
    ]
)

# ---------------------------------------------------
# Display First 5 Rows
# ---------------------------------------------------

print("\nFIRST 5 RECORDS")
print(df.head())

# ---------------------------------------------------
# Dataset Information
# ---------------------------------------------------

print("\nDATASET INFO")
print(df.info())

# ---------------------------------------------------
# Save CSV File
# ---------------------------------------------------

df.to_csv(
    "ecommerce_reviews.csv",
    index=False
)

print("\nData Saved Successfully")
```
